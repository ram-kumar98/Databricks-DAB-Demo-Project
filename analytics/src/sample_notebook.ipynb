# Default notebook

This default notebook is executed using a Lakeflow job as defined in resources/sample_job.job.yml.

```python
# Set default catalog and schema
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
```

```python
# Target table
target_table = "employee_details"
```

```python
full_table_name = f"{catalog}.{schema}.{target_table}"
```

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    IntegerType,
    StringType,
    DoubleType
)

data = [
    (1, "John", "IT", 75000.00),
    (2, "Alice", "HR", 65000.00),
    (3, "Bob", "Finance", 80000.00),
    (4, "David", "IT", 90000.00),
    (5, "Sarah", "Marketing", 70000.00)
]

schema = StructType([
    StructField("employee_id", IntegerType(), True),
    StructField("employee_name", StringType(), True),
    StructField("department", StringType(), True),
    StructField("salary", DoubleType(), True)
])

df = spark.createDataFrame(data, schema)

df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(full_table_name)
```
